```python
import yfinance as yf
import pandas as pd

tesla_ticker = yf.Ticker("TSLA")
tesla_data   = tesla_ticker.history(period="max").reset_index()
tesla_data.head()
```

```python
import requests
import pandas as pd

def get_quarterly_revenue(symbol:str, company_slug:str) -> pd.DataFrame:

    url      = f"https://www.macrotrends.net/stocks/charts/{symbol}/{company_slug}/revenue"
    headers  = {"User-Agent": "Mozilla/5.0"}
    html     = requests.get(url, headers=headers, timeout=30).text
    
    tables   = pd.read_html(html, flavor="bs4")
    
    revenue_table = None
    for tbl in tables:
        if tbl.shape[1] >= 2 and tbl.columns.map(str).str.contains("Revenue", case=False).any():
            revenue_table = tbl.copy()
            break
    
    if revenue_table is None:
        raise ValueError(f"No se encontró tabla de ingresos en {url}")

    revenue_table = revenue_table.rename(columns={revenue_table.columns[0]:"Date",
                                                  revenue_table.columns[1]:"Revenue"})
    revenue_table = revenue_table.dropna()
    revenue_table["Revenue"] = (revenue_table["Revenue"]
                                .astype(str)
                                .str.replace(r"[\$,]", "", regex=True)
                                .astype(float))
    return revenue_table
```

```python
tesla_revenue = get_quarterly_revenue("TSLA", "tesla")
tesla_revenue.tail()
```

```python
gme_ticker = yf.Ticker("GME")
gme_data   = gme_ticker.history(period="max").reset_index()
gme_data.head()
```

```python
gme_revenue = get_quarterly_revenue("GME", "gamestop")
gme_revenue.tail()
```

```python
# Función make_graph ─ Stock Price vs Revenue
import matplotlib.pyplot as plt

def make_graph(price_df:pd.DataFrame,
               revenue_df:pd.DataFrame,
               company_name:str) -> None:
    """Grafica precio de cierre y revenue en ejes gemelos."""
    fig, ax1 = plt.subplots(figsize=(12,6))

    # Precio de la acción
    ax1.plot(price_df["Date"], price_df["Close"], label="Precio cierre")
    ax1.set_xlabel("Fecha")
    ax1.set_ylabel("Precio (USD)")
    ax1.tick_params(axis='x', rotation=45)

    # Ingresos en eje secundario
    ax2 = ax1.twinx()
    ax2.plot(revenue_df["Date"], revenue_df["Revenue"],
             color="orange", label="Ingresos (M USD)", linestyle="--")
    ax2.set_ylabel("Ingresos (millones USD)")

    fig.suptitle(f"{company_name}: Precio vs Ingresos Trimestrales")
    fig.tight_layout()
    plt.show()
```

```python
make_graph(tesla_data, tesla_revenue, "Tesla")
```

```python
make_graph(gme_data, gme_revenue, "GameStop")
```
